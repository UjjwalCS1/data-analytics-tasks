#### Importing Libraraies

```python
import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
```

```python
df=pd.read_csv("marketing_campaign.csv", sep='\t')
```

```python
df.head()
```

### Data Cleaning Steps

```python
# Convert column names to lowercase and replace spaces with underscores
df.columns = df.columns.str.lower().str.replace(' ', '_')
```

#### Handling Missing Values

```python
# Check for missing values
print("Missing values before handling:")
display(df.isnull().sum())

```

```python
# Handle missing values in 'income' by filling with the mean
df['income'].fillna(df['income'].mean(), inplace=True)

print("\nMissing values after handling:")
display(df.isnull().sum())
```

#### Converting Data Types

```python
# Convert 'dt_customer' to datetime objects
df['dt_customer'] = pd.to_datetime(df['dt_customer'], format='%d-%m-%Y')

print("\nData types after conversion:")
display(df.info())
```

#### Removing Duplicate Rows

```python
initial_rows = df.shape[0]
# Check for and remove duplicate rows
df.drop_duplicates(inplace=True)

rows_after_deduplication = df.shape[0]
print(f"Number of duplicate rows removed: {initial_rows - rows_after_deduplication}")
```

#### Calculating Age and Handling Outliers

```python
import datetime

# Calculate 'age' from 'year_birth'
current_year = datetime.datetime.now().year
df['age'] = current_year - df['year_birth']

# Identify and handle outliers in 'age' (e.g., age > 100 or < 18)
# First, find a reasonable median age from valid entries
median_age = df['age'][(df['age'] > 18) & (df['age'] < 100)].median()

# Replace outliers with the calculated median age
df.loc[(df['age'] > 100) | (df['age'] < 18), 'age'] = median_age

print("\nAge distribution after handling outliers:")
display(df['age'].describe())
```

### Data Cleaning Summary

Here's a summary of the data cleaning steps performed:

1.  **Column Name Standardization**: Converted all column names to lowercase and replaced spaces with underscores.
2.  **Missing Value Handling**: Filled missing values in the `income` column with the mean income.
3.  **Date Conversion**: Converted the `dt_customer` column to datetime objects with the format `DD-MM-YYYY`.
4.  **Duplicate Removal**: Removed any duplicate rows from the dataset.
5.  **Age Calculation and Outlier Handling**: Calculated `age` from `year_birth` and handled outliers (age > 100 or < 18) by replacing them with the median age of valid entries.
